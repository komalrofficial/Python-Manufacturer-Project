VISULIZATION

Average Sales by Manufacturer

```python
filter1=df_cars['Manufacturer'].isin([ 'Dodge','Ford','Chevrolet','Mercedes-B','Toyota','Chrysler','Mitsubishi','Nissan'])
filter2=df_cars['Vehicle_type'].isin(['Passenger','Car'])
```

```python
filtered_df=df_cars[filter1 & filter2]
```

```python
import matplotlib.pyplot as plt

sales = filtered_df.groupby('Manufacturer')['Sales_in_thousands'].mean()

sales.plot(kind='bar', figsize=(8,5))

plt.title("Average Sales by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Average Sales (Thousands)")
plt.xticks(rotation=0)
plt.show()
```

BAR CHART FOR AVERAGE PRICE BY MANUFACTURER

```python
price = filtered_df.groupby('Manufacturer')['Price_in_thousands'].mean()

price.plot(kind='bar', figsize=(8,5))

plt.title("Average Price by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Average Price (Thousands)")
plt.xticks(rotation=0)
plt.show()
```

Grouped Bar Chart(Average Sales and Average Price by Manufacturer)

```python
import matplotlib.pyplot as plt

result = filtered_df.groupby('Manufacturer')[
    ['Sales_in_thousands', 'Price_in_thousands']
].mean()

result.plot(kind='bar', figsize=(10,5))

plt.title("Average Sales and Average Price by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Average (Thousands)")
plt.xticks(rotation=0)
plt.legend(['Average Sales', 'Average Price'])
plt.show()
```
